```python
#multivariate analysis using quality groups

df['quality_group'] = pd.cut(
    df['quality'],
    bins = [0,5,6,10],
    labels=['Low','Medium','High']
)
```

```python
y=df['quality']
```

```python
X=df.drop(columns=['quality','quality_group'],errors='ignore')
```

```python
print(X.columns)
print(X.dtypes)
```

```python
X_train,X_test,y_train, y_test = train_test_split(X,y,test_size = 0.30, random_state =57, stratify =y)
```

```python
print('X_train : ', X_train.shape)
print('X_test: ', X_test.shape)
print('y_train: ', y_train.shape)
print('y_test: ', y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(criterion= 'entropy',random_state =57)
```

```python
dt_model.fit(X_train,y_train)
```

```python
y_train_pred = dt_model.predict(X_train)
y_test_pred = dt_model.predict(X_test)
```

```python
train_accuracy = accuracy_score(y_train, y_train_pred)
print("Training Accuracy", train_accuracy)#overfitting
```

```python
test_accuracy = accuracy_score(y_test, y_test_pred)
print("Testing Accuracy", test_accuracy) #overfitting
```

```python
print("Precision: ",
      precision_score(y_test, y_test_pred, average ='weighted'))
print("Recall: ",
      recall_score(y_test, y_test_pred, average ='weighted'))
print("F1 Score: ",
      f1_score(y_test, y_test_pred, average ='weighted'))
```

```python
print(classification_report(y_test,y_test_pred))
```

```python
#Confusion Matrix
cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(8,6))

sns.heatmap(cm, annot =True, fmt='d', cmap='Blues')

plt.xlabel('Predicted Quality')
plt.ylabel('Actual Quality')
plt.title('Confusion Matrix - Decision Tree')

plt.show()
```

```python
print('Tree Depth', dt_model.get_depth())
print('Number Of Leaves: ', dt_model.get_n_leaves())
```

```python
plt.figure(figsize=(25, 15))

plot_tree(
    dt_model,
    feature_names = X.columns,
    class_names=[str(c) for c in sorted(y.unique())],
    filled =True,
    rounded =True,
    fontsize =8
)

plt.title("Decision Tree - Entropy")
plt.show()
```

```python
plt.figure(figsize=(25, 15))

plot_tree(
    dt_model,
    feature_names = X.columns,
    class_names=[str(c) for c in sorted(y.unique())],
    filled =True,
    rounded =True,
    fontsize =8,
    impurity = True, 
    precision=2,
    max_depth=3
)

plt.title("Decision Tree - Top 3 ")
plt.show()
```
